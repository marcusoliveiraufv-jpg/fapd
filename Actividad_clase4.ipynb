```python
import nycflights13
import pandas as pd
from nycflights13 import flights

# 1. Explorar los objetos incluidos en el paquete nycflights13 (filtrando atributos internos)
objetos_paquete = [obj for obj in dir(nycflights13) if not obj.startswith("_")]
print("Tablas y objetos disponibles en nycflights13:", objetos_paquete)

# Crear una copia explícita de trabajo para preservar el dataset original intacto
df = flights.copy()

# 2. Verificación de dimensiones exactas (336.776 filas x 19 columnas)
print(f"\nDimensiones exactas del DataFrame (filas, columnas): {df.shape}")

# Visualización de las primeras 5 filas
print("\n--- Primeras filas (df.head()) ---")
display(df.head())

# Resumen estructural completo
print("\n--- Resumen estructural (df.info()) ---")
df.info()

# Listado explícito de tipos de datos por variable
print("\n--- Tipos de datos por columna (df.dtypes) ---")
print(df.dtypes)
```

```python
# 1. Quantificação absoluta total de valores ausentes (NaN) em todo o DataFrame
total_nan = df.isna().sum().sum()
total_celdas = df.shape[0] * df.shape[1]
porcentagem_global_nan = round((total_nan / total_celdas) * 100, 2)

print(f"Total absoluto de valores NaN no dataset: {total_nan:,}")
print(f"Porcentagem global de incompletude: {porcentagem_global_nan}%")

# Detalhamento da contagem absoluta e percentual de NaN por coluna
resumo_nan = pd.DataFrame({
    "Valores_NaN": df.isna().sum(),
    "Porcentagem_NaN (%)": (df.isna().mean() * 100).round(2)
}).sort_values(by="Valores_NaN", ascending=False)

print("\n--- Auditoria de valores faltantes por variável ---")
display(resumo_nan[resumo_nan["Valores_NaN"] > 0])
```

```python
# 1. Filtrar los vuelos del 1 de enero y los vuelos de United Airlines (UA) con retraso >= 400 min
vuelos_1_enero = df[(df["month"] == 1) & (df["day"] == 1)].copy()
vuelos_ua_400 = df[(df["carrier"] == "UA") & (df["dep_delay"] >= 400)].copy()

print(f"1a. Total de vuelos el 1 de enero: {len(vuelos_1_enero)}")
display(vuelos_1_enero.head(3))

print(f"\n1b. Total de vuelos de UA con retraso >= 400 min: {len(vuelos_ua_400)}")
display(vuelos_ua_400)

# 2. Filtrar los vuelos de los dos primeros días de cada mes: operador OR (|) frente a .between(1, 2)
vuelos_dias_or = df[(df["day"] == 1) | (df["day"] == 2)].copy()
vuelos_dias_between = df[df["day"].between(1, 2)].copy()

print(f"\n2. Registros con operador '|': {len(vuelos_dias_or)} | Registros con .between(1, 2): {len(vuelos_dias_between)}")
print(f"¿Ambos métodos producen exactamente el mismo resultado?: {vuelos_dias_or.equals(vuelos_dias_between)}")

# 3. Filtrar los vuelos con destino a los aeropuertos de Houston (IAH y HOU) mediante .isin()
vuelos_houston = df[df["dest"].isin(["IAH", "HOU"])].copy()
print(f"\n3. Total de vuelos con destino a Houston (IAH/HOU): {len(vuelos_houston)}")
display(vuelos_houston[["year", "month", "day", "carrier", "flight", "origin", "dest"]].head(3))

# 4. Obtener los 3 vuelos con mayor retraso en salida (nlargest) y los 3 con menor retraso (nsmallest)
top3_mayor_retraso = df.nlargest(3, "dep_delay").copy()
top3_menor_retraso = df.nsmallest(3, "dep_delay").copy()

print("\n4a. Top 3 vuelos con MAYOR retraso en salida (nlargest):")
display(top3_mayor_retraso[["year", "month", "day", "carrier", "flight", "origin", "dest", "dep_delay"]])

print("\n4b. Top 3 vuelos con MENOR retraso en salida / más adelantados (nsmallest):")
display(top3_menor_retraso[["year", "month", "day", "carrier", "flight", "origin", "dest", "dep_delay"]])

# 5. Vuelos de la compañía UA con retraso en salida >= 400 minutos (vista detallada)
vuelos_ua_extremos = df[(df["carrier"] == "UA") & (df["dep_delay"] >= 400)].copy()
print(f"\n5. Detalle de los {len(vuelos_ua_extremos)} vuelos de UA con dep_delay >= 400 min:")
display(vuelos_ua_extremos[["year", "month", "day", "carrier", "flight", "origin", "dest", "dep_delay", "arr_delay"]])
```

```python
# 1. Ordenar los vuelos de menor a mayor retraso en la salida (ascending=True)
df_orden_asc = df.sort_values(by="dep_delay", ascending=True).copy()
print("1. Vuelos ordenados de menor a mayor retraso en la salida:")
display(df_orden_asc[["year", "month", "day", "carrier", "flight", "dep_delay"]].head(5))

# 2. Ordenar de mayor a menor retraso en salida y, ante empate, de menor a mayor retraso en llegada
df_orden_compuesto = df.sort_values(by=["dep_delay", "arr_delay"], ascending=[False, True]).copy()
print("\n2. Ordenación compuesta (dep_delay DESC, arr_delay ASC):")
display(df_orden_compuesto[["year", "month", "day", "carrier", "flight", "dep_delay", "arr_delay"]].head(5))

# 3. Crear subconjuntos de columnas por nombre, por rango (.loc) y por patrones de texto (.str)
sub_nombres = df[["year", "month", "day"]].copy()
sub_rango = df.loc[:, "year":"day"].copy()
sub_starts_a = df.loc[:, df.columns.str.startswith("a")].copy()
sub_ends_y = df.loc[:, df.columns.str.endswith("y")].copy()
sub_contains_ho = df.loc[:, df.columns.str.contains("ho")].copy()

print("\n3. Columnas seleccionadas según cada criterio:")
print("   - Por nombre explícito:", list(sub_nombres.columns))
print("   - Por rango (.loc[:, 'year':'day']):", list(sub_rango.columns))
print("   - Empiezan con 'a' (startswith):", list(sub_starts_a.columns))
print("   - Terminan con 'y' (endswith):", list(sub_ends_y.columns))
print("   - Contienen 'ho' (contains):", list(sub_contains_ho.columns))

# 4. Renombrado controlado sobre una copia de prueba (df_ren = df.copy())
df_ren = df.copy()
df_ren.rename(columns={"dep_time": "hora_de_salida"}, inplace=True)

print("\n4. Verificación de aislamiento tras el renombrado:")
print("   - ¿Existe 'hora_de_salida' en df_ren?:", "hora_de_salida" in df_ren.columns)
print("   - ¿El dataset original df conserva intacta 'dep_time'?:", "dep_time" in df.columns)
```

```python
# Trabajar sobre una copia explícita para preservar el dataset original auditado intacto
df_feat = df.copy()

# 1. Calcular la velocidad media del vuelo en metros por segundo (m/s)
# Conversión física: 1 milla = 1.609,34 metros | 1 minuto = 60 segundos
df_feat["velocidad_media"] = (df_feat["distance"] * 1609.34) / (df_feat["air_time"] * 60)

# 2. Crear la variable ganancia (minutos recuperados durante el trayecto)
# Fórmula: retraso en salida (dep_delay) - retraso en llegada (arr_delay)
df_feat["ganancia"] = df_feat["dep_delay"] - df_feat["arr_delay"]

# Inspección de las nuevas características creadas y sus estadísticos descriptivos
print("Dimensiones de df_feat (con nuevas variables):", df_feat.shape)
print("¿Dataset original df intacto (19 columnas)?:", df.shape)

display(df_feat[["carrier", "flight", "origin", "dest", "distance", "air_time", "velocidad_media", "dep_delay", "arr_delay", "ganancia"]].head(5))

print("\n--- Resumen estadístico de las nuevas variables ---")
display(df_feat[["velocidad_media", "ganancia"]].describe().round(2))
```

```python
# 1. Media y desviación estándar diaria de dep_delay agrupando por fecha (year, month, day)
resumen_diario = df.groupby(["year", "month", "day"]).agg(
    avg_dep_delay=("dep_delay", "mean"),
    std_dep_delay=("dep_delay", "std")
).reset_index()

print("1. Resumen diario de centralidad y dispersión del retraso en salida:")
display(resumen_diario.head())

# 2. Moda del retraso de salida por aerolínea (carrier) manejando posibles empates con .iloc[0]
moda_aerolinea = df.groupby("carrier").agg(
    moda_dep_delay=("dep_delay", lambda x: x.mode().iloc[0] if not x.mode().empty else None)
).reset_index().sort_values(by="moda_dep_delay", ascending=False)

print("\n2. Moda del retraso en salida por aerolínea (carrier):")
display(moda_aerolinea)

# 3. Hora del primer y último vuelo diario: valores extremos (.min/.max) vs posición ('first'/'last')
vuelos_diarios = df.groupby(["year", "month", "day"]).agg(
    primer_vuelo_min=("dep_time", "min"),
    ultimo_vuelo_max=("dep_time", "max"),
    primer_vuelo_pos=("dep_time", "first"),
    ultimo_vuelo_pos=("dep_time", "last")
).reset_index()

print("\n3. Comparativa del primer y último vuelo diario (extremos vs posición):")
display(vuelos_diarios.head())

# 4. Aeropuerto de destino (dest) que opera con mayor número de aerolíneas distintas (nunique)
aerolineas_por_destino = df.groupby("dest").agg(
    aerolineas=("carrier", "nunique")
).reset_index().sort_values(by="aerolineas", ascending=False)

print("\n4. Top 10 aeropuertos de destino con mayor diversidad de aerolíneas:")
display(aerolineas_por_destino.head(10))
```

```python
# Trabajar sobre una copia explícita para mantener intacto el DataFrame original df
df_cod = df.copy()

# 1. Comprobar las 16 aerolíneas únicas y aplicar Label Encoding con .cat.codes
aerolineas_unicas = df_cod["carrier"].unique()
print(f"1a. Total de aerolíneas únicas: {len(aerolineas_unicas)}")
print("    Códigos IATA:", aerolineas_unicas)

df_cod["carrier_code"] = df_cod["carrier"].astype("category").cat.codes
print(f"\n1b. Tipo de dato de 'carrier_code': {df_cod['carrier_code'].dtypes}")
print("    Primeras 5 filas de 'carrier_code':")
display(df_cod[["carrier", "carrier_code"]].head())

# 2. Aplicar codificación One-Hot (Dummy) sobre 'carrier' generando 16 columnas binarias 0/1
flights_onehot = pd.get_dummies(df, columns=["carrier"], dtype=int)
cols_onehot = [col for col in flights_onehot.columns if col.startswith("carrier_")]

print(f"\n2. Dimensiones de flights_onehot: {flights_onehot.shape} (se crearon {len(cols_onehot)} columnas binarias)")
display(flights_onehot[["year", "month", "day", "flight"] + cols_onehot].head())

# 3. Aplicar codificación manual personalizada mediante diccionario y .map()
mapa_carrier = {"UA": 0, "AA": 1, "B6": 2, "DL": 3}
df_cod["carrier_manual"] = df_cod["carrier"].map(mapa_carrier)

print("\n3. Resultado de codificación manual con .map() (las aerolíneas fuera del diccionario pasan a NaN):")
display(df_cod[["carrier", "carrier_manual"]].head(10))
print("   Conteo de valores generados (incluyendo NaN para categorías omitidas):")
print(df_cod["carrier_manual"].value_counts(dropna=False))
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# 1. Crear un subconjunto numérico sin valores nulos sobre una copia explícita
cols = ["dep_delay", "arr_delay", "air_time", "distance"]
df_num = df[cols].dropna().copy()

print(f"1. Dimensiones de df_num (sin nulos): {df_num.shape} (de {len(df)} filas originales)")
print("\n--- Estadísticos descriptivos originales (df_num) ---")
display(df_num.describe().round(2))

# 2. Estandarización Z-score (media = 0.00, desviación estándar = 1.00)
scaler_std = StandardScaler()
df_std = pd.DataFrame(
    scaler_std.fit_transform(df_num),
    columns=[c + "_std" for c in cols],
    index=df_num.index
)

print("\n2. Verificación de Estandarización Z-score (df_std):")
display(df_std.describe().round(2))

# 3. Normalización Min-Max (intervalo cerrado [0.0, 1.0])
scaler_minmax = MinMaxScaler()
df_norm = pd.DataFrame(
    scaler_minmax.fit_transform(df_num),
    columns=[c + "_norm" for c in cols],
    index=df_num.index
)

print("\n3. Verificación de Normalización Min-Max (df_norm):")
display(df_norm.describe().round(2))
```

```python
# Trabajar sobre una copia explícita para preservar el dataset original intacto
df_clean = df.copy()

# 1. Conversión robusta a tipo numérico con pd.to_numeric(..., errors='coerce')
df_clean["dep_delay"] = pd.to_numeric(df_clean["dep_delay"], errors="coerce")

# Demostración de robustez frente a texto inválido en una serie de prueba
serie_prueba = pd.Series(["15", "-5", "CANCELADO", None, "42.5"])
print("1. Prueba de pd.to_numeric con errors='coerce':")
print(pd.to_numeric(serie_prueba, errors="coerce").tolist())

# 2. Parseo de horas codificadas en HHMM (ej. 517 -> 05:17:00) a objetos de hora real
cols_datetime = ["dep_time", "sched_dep_time", "arr_time", "sched_arr_time"]
for col in cols_datetime:
    df_clean[col] = pd.to_datetime(
        df_clean[col].astype("Int64").astype(str).str.zfill(4),
        format="%H%M",
        errors="coerce"
    ).dt.time

print("\n2. Muestra de columnas horarias parseadas (HH:MM:SS):")
display(df_clean[cols_datetime].head())

# 3. Combinar columnas enteras de fecha y hora programada en datetime64[ns]
df_clean["datetime"] = pd.to_datetime(df_clean[["year", "month", "day", "hour", "minute"]])

print("\n3. Verificación de la nueva columna 'datetime':")
print("   Tipo de dato:", df_clean["datetime"].dtypes)
display(df_clean[["year", "month", "day", "hour", "minute", "datetime"]].head())
```
