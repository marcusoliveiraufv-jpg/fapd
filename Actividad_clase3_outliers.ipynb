```python
import os
import glob
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# Procura o arquivo CSV ignorando maiúsculas/minúsculas na pasta atual e subpastas
arquivos = [f for f in glob.glob("**/*.csv", recursive=True) if "outliers" in f.lower()]

if not arquivos:
    print(f"Arquivo não encontrado! Verifique se o CSV está na mesma pasta do notebook: {os.getcwd()}")
else:
    print(f"Carregando arquivo encontrado: {arquivos[0]}\n")
    df = pd.read_csv(arquivos[0])

    # 2. Estatísticas descritivas
    print(df[["edad", "ingreso_mensual", "tiempo_sesion_min"]].describe().round(2))

    # 3. Diagramas de caixa (Boxplots) independentes
    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df[["ingreso_mensual"]])
    plt.title("Ingreso Mensual")
    plt.show()

    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df[["edad"]])
    plt.title("Edad")
    plt.show()

    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df[["tiempo_sesion_min"]])
    plt.title("Tiempo de Sesión (min)")
    plt.show()
```
