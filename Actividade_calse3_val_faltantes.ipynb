```python
import pandas as pd
import numpy as np
from pathlib import Path

# 1. Localizar y leer el archivo DATA_Clientes_NaN.csv en el DataFrame original 'df'
archivo = next(Path.cwd().rglob("DATA_Clientes_NaN.csv"), None) or next(Path("/Users/marcusoliveira/Next/fapd").rglob("DATA_Clientes_NaN.csv"))
df = pd.read_csv(archivo)

# 2. Dimensiones exactas, primeros 8 registros e información técnica
print(f"Dimensiones del dataset: {df.shape[0]} filas y {df.shape[1]} columnas\n")

print("--- PRIMEROS 8 REGISTROS ---")
display(df.head(8))

print("--- INFORMACIÓN TÉCNICA (info) ---")
df.info()
```

```python
# 1. Máscara booleana de ausentes en las primeras 10 filas (True = NaN, False = dato válido)
print("--- MÁSCARA BOOLEANA DE NULOS (primeras 10 filas) ---")
display(df.isna().head(10))

# 2. Número exacto de valores NaN por variable y total global en todo el DataFrame
print("--- CANTIDAD DE VALORES NaN POR VARIABLE ---")
print(df.isna().sum())

total_celdas_nulas = df.isna().sum().sum()
print(f"\nTotal global de celdas ausentes en el DataFrame: {total_celdas_nulas}")

# 3. Porcentaje de incompletitud por columna
porcentaje_incompletitud = (df.isna().mean() * 100).round(2)
print("\n--- PORCENTAJE DE INCOMPLETITUD POR COLUMNA (%) ---")
print(porcentaje_incompletitud)
```

```python
# Respuesta al Ejercicio 1.2 (Paso 3)
print("""
Análisis de Incompletitud por Variable:

- Variable que supera el 50 % de valores ausentes: codigo_referido_antiguo (75,0 % de registros faltantes, al corresponder a una campaña heredada ya descatalogada).
- Variables con ausencia moderada (< 20 %): fecha_registro, ciudad, edad, ingresos_mensuales, gasto_mensual y bono_fidelidad (todas por debajo del 20 %, mientras que id_cliente presenta 0 % de ausentes al ser la clave primaria).
""")
```

```python
# 1. Simular borrado de todas las filas con al menos un valor ausente
df_sin_filas = df.dropna()
filas_sobreviven = len(df_sin_filas)
filas_originales = len(df)
porcentaje_perdido = ((filas_originales - filas_sobreviven) / filas_originales) * 100

print(f"Filas originales: {filas_originales}")
print(f"Filas que sobreviven tras df.dropna(): {filas_sobreviven}")
print(f"Porcentaje del volumen útil perdido: {porcentaje_perdido:.2f}%\n")

# 2. Simular eliminación de todas las columnas con al menos un NaN
df_sin_cols = df.dropna(axis=1)
print("--- COLUMNAS QUE QUEDAN TRAS df.dropna(axis=1) ---")
print(list(df_sin_cols.columns))
display(df_sin_cols.head())
print("""
Análisis Ejercicio 2.1 (Paso 2):
Aplicar dropna() de forma indiscriminada es inviable en este dataset porque:
- Por filas (axis=0): se pierde la gran mayoría de los clientes (superando ampliamente el umbral residual del 5 %), destruyendo el volumen útil de la muestra.
- Por columnas (axis=1): solo sobrevive 'id_cliente' (la clave primaria), eliminando todas las variables analíticas del negocio (edad, ciudad, ingresos_mensuales, gasto_mensual, bono_fidelidad y fecha_registro).
""")
```

```python
# 1. Crear df_limpio eliminando únicamente la variable con 80 % de nulos (notas_encuesta)
df_limpio = df.drop(["notas_encuesta"], axis=1)
print("Dimensiones de df_limpio:", df_limpio.shape)
print("Columnas actuales de df_limpio:", list(df_limpio.columns))

# 2. Practicar eliminación puntual de registros por etiqueta de índice sobre una copia de prueba
df_prueba = df_limpio.copy()
df_prueba = df_prueba.drop([3, 7])
print("\nDimensiones de df_prueba tras eliminar índices 3 y 7:", df_prueba.shape)
```

```python
# 1. Crear una copia de trabajo limpia desde df_limpio
df_imputado = df_limpio.copy()

# 2. Imputar valores ausentes con valor fijo 0 (en tu CSV la columna es 'compras_ultimo_mes')
df_imputado["compras_ultimo_mes"] = df_imputado["compras_ultimo_mes"].fillna(0)

print("Valores nulos restantes en 'compras_ultimo_mes':", df_imputado["compras_ultimo_mes"].isna().sum())
```

```python
# 1. Comparar el efecto de ffill() frente a bfill() en registros ordenados por id_cliente
comparacion_fechas = pd.DataFrame({
    "original": df_imputado["fecha_registro"],
    "ffill": df_imputado["fecha_registro"].ffill(),
    "bfill": df_imputado["fecha_registro"].bfill()
})

# Mostrar las filas que originalmente tenían NaN (y los índices de referencia disponibles)
indices_nulos_fecha = df_imputado[df_imputado["fecha_registro"].isna()].index
print("--- COMPARACIÓN FFILL VS BFILL EN FILAS CON FECHA AUSENTE ---")
display(comparacion_fechas.loc[indices_nulos_fecha])

# 2. Aplicar definitivamente la propagación hacia adelante (ffill) para mantener coherencia temporal
df_imputado["fecha_registro"] = df_imputado["fecha_registro"].ffill()
print("Valores nulos restantes en 'fecha_registro':", df_imputado["fecha_registro"].isna().sum())
```

```python
# 1. Variable numérica simétrica (edad): imputar con la media
media_edad = df_imputado["edad"].mean().round(2)
print(f"Promedio de edad: {media_edad}")
df_imputado["edad"] = df_imputado["edad"].fillna(df_imputado["edad"].mean())

# 2. Variables numéricas con valores atípicos (ingresos_mensuales y gasto_mensual): comparar media vs mediana
media_ingresos = df_imputado["ingresos_mensuales"].mean().round(2)
mediana_ingresos = df_imputado["ingresos_mensuales"].median().round(2)
print(f"Ingresos mensuales -> Media: {media_ingresos} € | Mediana: {mediana_ingresos} € (robusta frente a outliers)")

# Imputar ambas variables con su respectiva mediana
df_imputado["ingresos_mensuales"] = df_imputado["ingresos_mensuales"].fillna(df_imputado["ingresos_mensuales"].median())
df_imputado["gasto_mensual"] = df_imputado["gasto_mensual"].fillna(df_imputado["gasto_mensual"].median())

# 3. Variable categórica (ciudad): imputar con la moda
moda_ciudad = df_imputado["ciudad"].mode()[0]
print(f"Ciudad más frecuente (moda): {moda_ciudad}")
df_imputado["ciudad"] = df_imputado["ciudad"].fillna(moda_ciudad)

# 4. Certificar el 100 % de completitud (0 NaN en todas las variables) y 100 % de volumen útil
print("\n--- AUDITORÍA FINAL DE NULOS (df_imputado.isna().sum()) ---")
print(df_imputado.isna().sum())

print("\n--- INFORMACIÓN FINAL DEL DATASET IMPUTADO ---")
df_imputado.info()
```
