```python
import pandas as pd

# 1. Leer el archivo CSV (ajustado a la ruta de tu carpeta archivos)
ruta = "archivos/DATA_Clase03_Archivos_Actividades/DATA_Estudiantes.csv"
df_estudiantes = pd.read_csv(ruta)

# 2. Observar las 5 primeras filas
print("--- 5 PRIMERAS FILAS ---")
display(df_estudiantes.head())

# Observar las 5 últimas filas
print("--- 5 ÚLTIMAS FILAS ---")
display(df_estudiantes.tail())
```

```python
# 1. Tamaño exacto del DataFrame (filas, columnas)
filas, columnas = df_estudiantes.shape
print(f"El DataFrame tiene un total de {filas} filas (registros) y {columnas} columnas (variables).\n")

# 2. Resumen técnico del objeto
print("--- INFORMACIÓN ESTRUCTURAL ---")
df_estudiantes.info()
```

```python
# 1. Inspeccionar el tipo de dato de cada columna y listar nombres de variables
print("--- TIPOS DE DATOS (dtypes) ---")
print(df_estudiantes.dtypes)

print("\n--- LISTA DE COLUMNAS (columns) ---")
print(df_estudiantes.columns)

# 2. Separar columnas numéricas y categóricas
cols_numericas = df_estudiantes.select_dtypes(include="number").columns
df_categoricas = df_estudiantes.select_dtypes(include=["object", "bool"])

print("\n--- COLUMNAS NUMÉRICAS (con .columns -> devuelve un Index con los nombres) ---")
print(cols_numericas)

print("\n--- DATAFRAME CATEGÓRICO (sin .columns -> devuelve un DataFrame con los datos) ---")
display(df_categoricas.head())
```

```python
# 4. Estadísticos descriptivos de todas las variables numéricas
print("--- ESTADÍSTICOS DESCRIPTIVOS GENERALES ---")
display(df_estudiantes.describe())

# Resumen específico de las calificaciones redondeado a 2 decimales
print("--- RESUMEN DE CALIFICACIONES (nota_parcial y nota_final) ---")
display(df_estudiantes[["nota_parcial", "nota_final"]].describe().round(2))
```

```python
# 1. Conteo absoluto de valores faltantes por columna
nulos_total = df_estudiantes.isnull().sum()
print("--- CANTIDAD DE VALORES FALTANTES POR COLUMNA ---")
print(nulos_total)

# Porcentaje de valores faltantes respecto al total de filas
nulos_porcentaje = (df_estudiantes.isnull().mean() * 100).round(2)
print("\n--- PORCENTAJE DE VALORES FALTANTES (%) ---")
print(nulos_porcentaje)
```

```python
# 1. Número total de registros exactamente duplicados
total_duplicados = df_estudiantes.duplicated().sum()
print(f"Total de filas exactamente duplicadas: {total_duplicados}\n")

# 2. Visualizar todas las filas involucradas en la duplicidad (original + copia)
print("--- FILAS INVOLUCRADAS EN DUPLICIDAD (keep=False) ---")
display(df_estudiantes[df_estudiantes.duplicated(keep=False)])
```
