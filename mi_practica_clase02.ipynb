```python
# 1. Creación de variables siguiendo las convenciones de Python (snake_case)
nombre = "Marcus"
edad = 28
altura = 1.78
es_estudiante = True

# 2. Impresión del valor y la clase de cada objeto con print() y type()
print("nombre:", nombre, "| Clase:", type(nombre))
print("edad:", edad, "| Clase:", type(edad))
print("altura:", altura, "| Clase:", type(altura))
print("es_estudiante:", es_estudiante, "| Clase:", type(es_estudiante))

# 3. Inspección del espacio de nombres activo (solo variables creadas)
print("\nVariables en el espacio de nombres activo:")
print([var for var in dir() if not var.startswith('_') and var not in ('In', 'Out', 'exit', 'get_ipython', 'open', 'quit')])
```

```python
import math

# Fase 1 — Definición
distancia = 384400       # int: km Tierra-Luna
velocidad = 28000.5      # float: km/h
dias_mision = 7          # int

# Fase 2 — Cálculos y Redondeo
tiempo_viaje = distancia / velocidad
print("Tiempo de viaje (horas):", round(tiempo_viaje, 1))

# Fase 3 — Uso del módulo math
raiz_distancia = math.sqrt(distancia)
print("Raíz cuadrada de la distancia:", raiz_distancia)
```

```python
# ==============================================================================
# EJERCICIO 3.1: Listas — Modificación In-place, Indexación y Slicing
# ==============================================================================
print("--- EJERCICIO 3.1 ---")
frutas = ["manzana", "banana", "cereza", "dátil"]
print("Tercer elemento (índice 2):", frutas[2])
print("Último elemento (índice -1):", frutas[-1])

frutas.append("mango")
frutas.insert(1, "pera")
extraido = frutas.pop(2)

print("Elemento extraído:", extraido)
print("Lista final de frutas:", frutas)

numeros = [10, 20, 30, 40, 50, 60]
sub_lista_a = numeros[1:4]
print("sub_lista_a:", sub_lista_a)


# ==============================================================================
# EJERCICIO 3.2: Tuplas y Cadenas de Texto (Secuencias Inmutables)
# ==============================================================================
print("\n--- EJERCICIO 3.2 ---")
datos = (1, 5, 2, 5, 8, 5, 3)
print("Apariciones del 5:", datos.count(5))
print("Longitud:", len(datos))
print("Mínimo:", min(datos))
print("Máximo:", max(datos))

texto = " Programar en Python es divertido "
texto_limpio = texto.strip()
texto_lista = texto_limpio.split()
palabra_python = texto_limpio[13:19]

print("Texto limpio:", repr(texto_limpio))
print("texto_lista:", texto_lista)
print("Palabra extraída por slicing:", palabra_python)


# ==============================================================================
# EJERCICIO 3.3: Conjuntos (set) y Eliminación de Duplicados
# ==============================================================================
print("\n--- EJERCICIO 3.3 ---")
numeros_duplicados = [1, 2, 2, 3, 4, 4, 4]
numeros_set = set(numeros_duplicados)

numeros_set.add(5)
numeros_set.remove(2)

print("Conjunto resultante:", numeros_set)
```

```python
import pandas as pd

# ==============================================================================
# EJERCICIO 4.1: Creación, Acceso y Actualización de Diccionarios
# ==============================================================================
print("--- EJERCICIO 4.1 ---")
info_libro = {
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "año": 1967
}

print("Autor:", info_libro["autor"])

info_libro["paginas"] = 417
info_libro["año"] = 1970

print("Diccionario actualizado:", info_libro)


# ==============================================================================
# EJERCICIO 4.2: Creación de DataFrame, Selección (.iloc) y Filtrado (.isin)
# ==============================================================================
print("\n--- EJERCICIO 4.2 ---")
datos_empleados = {
    "ID": [101, 102, 103, 104],
    "Nombre": ["Ana", "Luis", "Marta", "Carlos"],
    "Departamento": ["Ventas", "IT", "Ventas", "Marketing"],
    "Ciudad": ["Madrid", "Valencia", "Madrid", "Sevilla"],
    "Salario": [50000, 65000, 52000, 48000]
}
df_empleados = pd.DataFrame(datos_empleados)

print("Columnas:", df_empleados.columns)
print("\nColumna Salario:\n", df_empleados["Salario"])

print("\nPrimeras 2 filas y 3 columnas (.iloc):\n", df_empleados.iloc[0:2, 0:3])

df_ventas = df_empleados[df_empleados["Departamento"] == "Ventas"]
print("\nFiltrado Departamento == 'Ventas':\n", df_ventas)

df_ciudades = df_empleados[df_empleados["Ciudad"].isin(["Madrid", "Sevilla"])]
print("\nFiltrado Ciudad en ['Madrid', 'Sevilla']:\n", df_ciudades)


# ==============================================================================
# EJERCICIO 4.3: Combinación de DataFrames (pd.concat y pd.merge)
# ==============================================================================
print("\n--- EJERCICIO 4.3 ---")
df_nuevo = pd.DataFrame({
    "ID": [105],
    "Nombre": ["Elena"],
    "Departamento": ["BI"],
    "Ciudad": ["Bilbao"],
    "Salario": [58000]
})

df_empleados_concat = pd.concat([df_empleados, df_nuevo], ignore_index=True)
print("DataFrame concatenado (pd.concat):\n", df_empleados_concat)

df_proyectos = pd.DataFrame({
    "ID": [101, 103, 106],
    "Proyecto": ["IA Generativa", "ETL Cloud", "Data Lake"]
})

merge_inner = pd.merge(df_empleados, df_proyectos, on="ID", how="inner")
print("\nMerge INNER (solo IDs coincidentes: 101 y 103):\n", merge_inner)

merge_left = pd.merge(df_empleados, df_proyectos, on="ID", how="left")
print("\nMerge LEFT (todos los de df_empleados; sin proyecto aparecen como NaN):\n", merge_left)
```
